```python
import csv

def evaluate_student(name, attendance, grade, login_correct, safety_gear):
    eligible = attendance >= 75
    passed = grade >= 75
    bonus = attendance >= 90 and grade >= 85
    lab_access = safety_gear

    return {
        "Name": name,
        "Attendance": attendance,
        "Grade": grade,
        "Eligible": eligible,
        "Passed": passed,
        "Bonus": bonus,
        "LabAccess": lab_access
    }

students = [
    evaluate_student("Student A", 95, 90, True, True),
    evaluate_student("Student B", 70, 80, True, False),
    evaluate_student("Student C", 85, 74, False, True)
]

with open("logic_results.csv", "w", newline="") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=["Name", "Attendance", "Grade", "Eligible", "Passed", "Bonus", "LabAccess"]
    )
    writer.writeheader()
    writer.writerows(students)

print("Results saved to logic_results.csv")
```
